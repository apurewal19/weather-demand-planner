```python
import pandas as pd
import os
from sqlalchemy import create_engine
from dotenv import load_dotenv
```

```python
load_dotenv()

mysql_user = os.getenv("MYSQL_USER")
mysql_password = os.getenv("MYSQL_PASSWORD")
mysql_host = os.getenv("MYSQL_HOST")
mysql_db = os.getenv("MYSQL_DB")

engine = create_engine(f'mysql+mysqlconnector://{mysql_user}:{mysql_password}@{mysql_host}/{mysql_db}')
```

```python
rain_df = pd.read_sql('SELECT * FROM joined_rain', engine)
cold_df = pd.read_sql('SELECT * FROM joined_cold', engine)
warm_df = pd.read_sql('SELECT * FROM joined_warm', engine)

print(rain_df.shape)
print(cold_df.shape)
print(warm_df.shape)
```

```python
rain_df[['total_rain', 'rain_jacket', 'waterproof_boots', 'rain_pants', 'umbrella', 'rubber_boots']].corr()
```

```python
cold_df[['average_temp_mean', 'insulated_jacket', 'base_layer', 'winter_gloves', 'wool_socks', 'beanie']].corr()
```

```python
warm_df[['average_temp_max', 'hiking_boots', 'sandals', 'light_jacket', 'sun_hat', 'shorts']].corr()
```

```python
rain_df['total_rain_lag1'] = rain_df['total_rain'].shift(1)
rain_df['total_rain_lag2'] = rain_df['total_rain'].shift(2)

rain_df[['total_rain', 'total_rain_lag1', 'total_rain_lag2', 
         'rain_jacket', 'waterproof_boots', 'rain_pants', 'umbrella', 'rubber_boots']].corr()
```

```python
cold_df['average_temp_mean_lag1'] = cold_df['average_temp_mean'].shift(1)
cold_df['average_temp_mean_lag2'] = cold_df['average_temp_mean'].shift(2)

cold_df[['average_temp_mean', 'average_temp_mean_lag1', 'average_temp_mean_lag2',
         'insulated_jacket', 'base_layer', 'winter_gloves', 'wool_socks', 'beanie']].corr()
```

```python
warm_df['average_temp_max_lag1'] = warm_df['average_temp_max'].shift(1)
warm_df['average_temp_max_lag2'] = warm_df['average_temp_max'].shift(2)

warm_df[['average_temp_max', 'average_temp_max_lag1', 'average_temp_max_lag2',
         'hiking_boots', 'sandals', 'light_jacket', 'sun_hat', 'shorts']].corr()
```

```python
rain_corr = rain_df[['total_rain', 'rain_jacket', 'waterproof_boots', 'rain_pants', 'umbrella', 'rubber_boots']].corr()
cold_corr = cold_df[['average_temp_mean', 'insulated_jacket', 'base_layer', 'winter_gloves', 'wool_socks', 'beanie']].corr()
warm_corr = warm_df[['average_temp_max', 'hiking_boots', 'sandals', 'light_jacket', 'sun_hat', 'shorts']].corr()

# Pull just the weather-driver row from each — that's the column of correlations you actually care about
rain_summary = rain_corr.loc['total_rain'].drop('total_rain').reset_index()
rain_summary.columns = ['keyword', 'correlation']
rain_summary['category'] = 'Rain'
rain_summary['weather_driver'] = 'total_rain'

cold_summary = cold_corr.loc['average_temp_mean'].drop('average_temp_mean').reset_index()
cold_summary.columns = ['keyword', 'correlation']
cold_summary['category'] = 'Cold'
cold_summary['weather_driver'] = 'average_temp_mean'

warm_summary = warm_corr.loc['average_temp_max'].drop('average_temp_max').reset_index()
warm_summary.columns = ['keyword', 'correlation']
warm_summary['category'] = 'Warm'
warm_summary['weather_driver'] = 'average_temp_max'

# Combine all three into one table
summary_df = pd.concat([rain_summary, cold_summary, warm_summary], ignore_index=True)
summary_df['best_timing'] = 'Same-week'
summary_df = summary_df.round(2)
```

```python
summary_df.to_csv('demand_summary.csv', index=False)
```
